```python
import pandas as pd
from sklearn.cluster import KMeans
```

```python
df = pd.DataFrame({
    'age':    [22, 25, 47, 52, 46, 56, 35, 28, 40, 60],
    'salary': [20000, 25000, 50000, 60000, 55000, 70000, 40000, 30000, 45000, 80000]
})

model=KMeans(n_clusters=3)
model.fit(df)
```

```python
df['group']=model.labels_
df
```

```python
import matplotlib.pyplot as plt
colors=['red','green','blue']

for i in range(3):
  group=df[df['group']==i]
  plt.scatter(group['age'],group['salary'],
              color=colors[i],
              label=f'group{i+1}',
              s=100
              )
  plt.xlabel('Age')
  plt.ylabel('salary')
  plt.title('Customer Groups')
  plt.legend()
  plt.show()
```
