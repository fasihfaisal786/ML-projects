Model Training and Prediction

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

df = pd.DataFrame({
    'area':     [500, 800, 1000, 1200, 1500, 1800, 2000, 2200, 2500, 3000],
    'bedrooms': [1,   2,   2,    3,    3,    3,    4,    4,    5,    5],
    'price':    [10000, 16000, 20000, 24000, 30000, 36000, 40000, 44000, 50000, 60000]
})

X = df[['area', 'bedrooms']]
Y = df['price']

model = LinearRegression()
model.fit(X, Y)
score = r2_score(Y, model.predict(X))
print(f"Model Score: {score}")

prediction = model.predict([[2000, 4]])
print(f"2000 sqft, 4 bedroom ka price: {prediction[0]}")
```
